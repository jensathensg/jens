```r
# Part A

download.file(
  "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv",
  "penguins.csv",
  mode = "wb"
)

penguins <- read.csv("penguins.csv")

head(penguins)

str(penguins)

summary(penguins)

colSums(is.na(penguins))

penguins_clean <- subset(
  penguins,
  !is.na(body_mass_g) &
  !is.na(species) &
  !is.na(sex)
)

penguins_clean$species <- factor(penguins_clean$species)
penguins_clean$sex <- factor(penguins_clean$sex)

str(penguins_clean)

nrow(penguins_clean)
```

```r
#  Part B

summary(penguins_clean$body_mass_g)
sd(penguins_clean$body_mass_g)

aggregate(body_mass_g ~ species, penguins_clean, mean)
aggregate(body_mass_g ~ sex, penguins_clean, mean)
aggregate(body_mass_g ~ species + sex, penguins_clean, mean)

boxplot(body_mass_g ~ species, data = penguins_clean, col = "lightblue")
boxplot(body_mass_g ~ sex, data = penguins_clean, col = "lightpink")

interaction.plot(penguins_clean$species, penguins_clean$sex,
                 penguins_clean$body_mass_g, mean, type = "b")
```

1. Which species appears to have the highest average body mass?

Answer: The gentoo penguins have the highest average body mass.

2. Which sex appears to have the higher average body mass?

Answer: Male penguins tends to have the higher average body mass.

3. Do the differences between males and females appear similar across species?

Answer: No. The difference between male and female body mass varies across species. This conclude that the relationship between sex and body mass may depend on species.

```r
# Part C

penguins_clean$species <- relevel(
  penguins_clean$species,
  ref = "Adelie"
)

oneway_model <- aov(
  body_mass_g ~ species,
  data = penguins_clean
)

summary(oneway_model)
```

The mean of body mass is equal for all the three penguin species and at least one species has a different mean body mass.

$$ H_0: \mu_{_\text{Adelie}}=\mu_{\text{Chinstrap}}=\mu{\text{Gentoo}}$$

```r
# Part D

dummy_model <- lm(
  body_mass_g ~ species,
  data = penguins_clean
)

summary(dummy_model)

aggregate(
  body_mass_g ~ species,
  data = penguins_clean,
  FUN = mean
)
```

The regression model was used to determine how penguin body mass differs among species by comparing each species with a reference category. **Adelie** penguins were used as the reference category in this analysis. The dummy variables are the **Chinstrap** penguins and **Gentoo** penguins.

The estimated regression equation is $$Y_i =\beta_0 + \beta_1 D_{Chinstrap} + \beta_2 D_{Gentoo} + \epsilon_i$$

The body mass of Adelie penguins is approximately **3,706 grams**. The body mass of Chinstrap penguins are **27 grams** higher than that of Adelie penguins. While the Gentoo penguins are approximately **1,386 grams** higher than that of Adelie penguins.

The results shows that Gentoo penguins have greater average body mass than Adelie penguins, while the difference between Chinstrap and Adelie is relatively small. This shows how each each species differs from one another.



```r
# Part E

twoway_model <- aov(
  body_mass_g ~ species * sex,
  data = penguins_clean
)

summary(twoway_model)
```

1. Species Main Effect
*   $H_0$: The mean body mass are equal among species.
*   $H_1$: At least one species has a different mean body mass.
Species has a significant effect on body mass if the p-value is below 0.05.

2. Sex Main Effect
*   $H_0$: The mean body mass are equal to both males and females.
*   $H_1$: The mean body mass differs between males and females.
Sex has a significant effect on body mass if its p-value is below 0.05. Thus, males have a higher average body mass overall.

3. Species x Sex Interaction
*   $H_0$: There is no interaction between species and sex.
*   $H_1$: Thre is an interaction between the species and sex.
The difference between malea nd female body mass changes depending on the species.






```r
# Part F

interaction.plot(
  x.factor = penguins_clean$species,
  trace.factor = penguins_clean$sex,
  response = penguins_clean$body_mass_g,
  fun = mean,
  type = "b",
  xlab = "Penguin Species",
  ylab = "Mean Body Mass (g)",
  trace.label = "Sex",
  main = "Interaction Between Species and Sex"
)

summary(twoway_model)
```

a. Are the profiles apprixmately parallel?
Answer: No, it is not perfectly parallel to each other.

b. If they are non-parallel, what does this suggest?
Answer: It suggests that the difference in body mass between males and females depends on the penguin species.

c. Is the Species x Sex interaction statistically significant?
Answer: Yes. It is statistically significant at the 5% significance level $(p<0.05)$.

e. Explain the interaction in practicla terms?
Answer: The male pengins tend to be heavier than the female penguins in all three species. However, based on the observation the size of this difference varies by species. The difference is smallest among Chinstrap penguins and largest among Gentoo penguins.

```r
# Part G

full_model <- lm(
  body_mass_g ~ species * sex,
  data = penguins_clean
)

summary(full_model)

anova(full_model)
```

The **intercept** is the estimated mean body mass of the reference group, which is the female Adelie penguins. The **species coefficient** is the differences in mean body mass between the species including Chinstrap, Gentoo, and Adelie females. While the sex coefficient is the difference between the male and femeale body mass for the reference species, which is the Adelie. The interaction are important because the sex difference is not the same for every species.
