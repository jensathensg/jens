```r
advertising <- read.csv(
  "Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv"
)

model <- lm(sales ~ TV, data = advertising)

r <- cor(advertising$TV, advertising$sales)

r_squared <- r^2

model_r_squared <- summary(model)$r.squared

correlation_table <- data.frame(
  Measure = c(
    "Correlation Coefficient (r)",
    "Coefficient of Determination (r²)",
    "Model R²"
  ),
  Value = round(c(r, r_squared, model_r_squared), 4)
)

print(correlation_table, row.names = FALSE)
```
