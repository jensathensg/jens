```r
advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

head(advertising)

summary(advertising)

# Scatterplot
plot(advertising$TV,
     advertising$sales,
     main = "TV Advertising vs Sales",
     xlab = "TV Advertising",
     ylab = "Sales",
     col = "blue",
     pch = 19)

# Correlation
cor(advertising$TV, advertising$sales)

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

# Coefficients
intercept <- coef(model)[1]
slope <- coef(model)[2]

results <- data.frame(
  Term = c("Intercept", "Slope (TV)"),
  Value = c(intercept, slope)
)

results

# Regression Equation
equation <- data.frame(
  Equation = paste(
    "Sales =",
    round(intercept, 4),
    "+",
    round(slope, 4),
    "* TV"
  )
)

equation

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

advertising$Predicted_Sales <- fitted(model)

advertising$Residual <- residuals(model)

head(advertising[, c(
  "TV",
  "sales",
  "Predicted_Sales",
  "Residual"
)], 10)

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

plot(
  model,
  which = 1,
  main = "Residuals vs. Fitted Values"
)

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

summary(model)$sigma

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

y <- advertising$sales

yhat <- fitted(model)

TSS <- sum((y - mean(y))^2)

RSS <- sum((y - yhat)^2)

RegSS <- TSS - RSS

sums_of_squares <- data.frame(
  Measure = c(
    "TSS",
    "RSS",
    "RegSS",
    "RegSS + RSS"
  ),
  Value = round(
    c(
      TSS,
      RSS,
      RegSS,
      RegSS + RSS
    ),
    4
  )
)

print(sums_of_squares, row.names = FALSE)

advertising <- read.csv("Advertising - Advertising - Advertising - Advertising - Advertising - Advertising (1).csv")

model <- lm(sales ~ TV, data = advertising)

r <- cor(advertising$TV, advertising$sales)

r_squared <- r^2

model_r_squared <- summary(model)$r.squared

correlation_table <- data.frame(
  Measure = c(
    "Correlation Coefficient (r)",
    "Coefficient of Determination (r²)",
    "Model R²"
  ),
  Value = round(
    c(
      r,
      r_squared,
      model_r_squared
    ),
    4
  )
)

print(correlation_table, row.names = FALSE)
```
